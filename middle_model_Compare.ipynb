```python
import numpy as np

# 파일 불러오기
EfficientNetb0_acc= np.load('0/EfficientNetb0/vit_acc.npy')
EfficientNetb0_loss= np.load('0/EfficientNetb0/vit_loss.npy')

MobileNet_acc= np.load('0/MobileNet/MobileNet_accuracy.npy')
MobileNet_loss= np.load('0/MobileNet/MobileNet_accuracy.npy')

ResNet18_acc= np.load('0/ResNet18/ResNet18_acc.npy')
ResNet18_loss= np.load('0/ResNet18/ResNet18_loss.npy')

VGG16_acc= np.load('0/VGG16/acc1.npy')
VGG16_loss= np.load('0/VGG16/loss1.npy')

ViT_acc= np.load('0/vit/vit_acc1.npy')
ViT_loss= np.load('0/vit/vit_loss1.npy')

YOLO_acc= np.load('0/YOLO11ncls/YOLO11n-cls_acc_values.npy')
YOLO_loss= np.load('0/YOLO11ncls/YOLO11n-cls_loss_values.npy')
```

```python
print(VGG16_acc,VGG16_loss)
```

```python
import matplotlib.pyplot as plt

# 그래프 설정 (1행 2열)
plt.figure(figsize=(15, 6))

# 1. Accuracy 비교 그래프
plt.subplot(1, 2, 1)
plt.plot(EfficientNetb0_acc, label='EfficientNetB0')
plt.plot(MobileNet_acc, label='MobileNet')
plt.plot(ResNet18_acc, label='ResNet18')
plt.plot(VGG16_acc, label='VGG16')
# plt.plot(ViT_acc, label='Vision Transformer')
plt.plot(YOLO_acc, label='YOLO11n-cls')
plt.title('Model Accuracy Comparison')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

# 2. Loss 비교 그래프
plt.subplot(1, 2, 2)
plt.plot(EfficientNetb0_loss, label='EfficientNetB0')
plt.plot(MobileNet_loss, label='MobileNet')
plt.plot(ResNet18_loss, label='ResNet18')
plt.plot(VGG16_loss, label='VGG16')
# plt.plot(ViT_loss, label='Vision Transformer')
plt.plot(YOLO_loss, label='YOLO11n-cls')
plt.title('Model Loss Comparison')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

print("--- Final Accuracy Summary ---")
models = ['EfficientNetB0', 'MobileNet', 'ResNet18', 'VGG16', 'ViT', 'YOLO11n']

# 각 배열의 마지막 값을 가져오되, 배열일 경우 첫 번째 요소만 선택([0])하거나 평균을 냄
accs = [
    np.mean(EfficientNetb0_acc[-1]), 
    np.mean(MobileNet_acc[-1]), 
    np.mean(ResNet18_acc), 
    np.mean(VGG16_acc[-1]), 
    np.mean(ViT_acc[-1]), 
    np.mean(YOLO_acc[-1])
]

for model, acc in zip(models, accs):
    print(f"{model}: {float(acc):.4f}")
```
