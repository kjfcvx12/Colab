```python
import glob

img1=glob.glob('Pistol_Data/export/images/*.jpg')
```

```python
import os
import shutil
```

```python
# 이미지 개수를 반으로 줄여서
reduce_len=img1[::2]
reduce_dir='Pistol_Data/export/images/reduce_images'

# 새로운 폴더를 생성해서
os.makedirs(reduce_dir, exist_ok=True)

# 이미지를 새폴더에 복사
for img1 in reduce_len:
    shutil.copy(img1, os.path.join(reduce_dir, os.path.basename(img1)))
    print(img1)

print(len(reduce_len))
```

```python
# 학습, 검증, 테스트
from sklearn.model_selection import train_test_split

train_img, valid_img=train_test_split(reduce_len, test_size=0.2, random_state=1)

print(len(train_img), len(valid_img))
```
