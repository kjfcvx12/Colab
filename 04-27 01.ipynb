```python
# 임베딩(사람이 쓰는 언어를 컴퓨터가 계산할 수 있도록 수치화(벡터화)한 기술)
# 단어의 의미와 관계를 담은 좌표로 변환해야함

# BOW(Bag of Words)
sentences=['I love my dog',
           'I love my cat',
           'You love my dog and my cat']
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# CountVectorizer : 텍스트 수치화 과정(토큰 횟수 행렬로 변환하는 도구)
cv=CountVectorizer()
```

```python
# sentences 리스트를 훓으면서 중복제거하고 해당단어 몇번 나왔는지 카운트해서 숫자 행렬로 변환
features=cv.fit_transform(sentences)
features.toarray()
```

```python
# 학습된 단어들을 인덱스 순서대로
f_name=cv.get_feature_names_out()
f_name
```

```python
import pandas as pd
df=pd.DataFrame(features.toarray(), columns=f_name)
df
```
