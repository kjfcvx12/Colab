```python
from transformers import pipeline
```

```python
# pipeline(입력한 문장 -> 숫자로 벡터화/ 데이터를 신경망에 통과시켜 예측값 계산/ 출력값 -> 우리가 읽을 수 있게 바꿈)
# 입력처리/추론/결과 정제함
```

```python
translator=pipeline("translation", model="Helsinki-NLP/opus-mt-ko-en")
translator('내일은 쉬는 날'), translator('열심히 해야지')
```

```python
cls=pipeline('sentiment-analysis')
```

```python
cls('I am very happy')
```

```python
# sentiment-analysis -> 기본모델이 영어
# distilbert-base-uncased-finetuned-sst-2-english
cls2=pipeline('sentiment-analysis')
result=cls2(['I am very happy', 'I hate you'])
print(result)
```

```python
review=pipeline('sentiment-analysis', model='sangrimlee/bert-base-multilingual-cased-nsmc')
print(review('오늘 슬퍼서 소주 한잔 해야지'))
print(review('오늘 너무 즐거운 날이다. 행복하다'))
print(review('감사함을 느낍니다'))
print(review('별로 좋은 기억이 없다'))
```

```python
model='nlptown/bert-base-multilingual-uncased-sentiment'
kor_classfier=pipeline('sentiment-analysis', model=model)
```

```python
results=kor_classfier(['다시는 보고 싶지 않은 짜증나는 영화',
                       '역대급 재밌음',
                       '아주 재미있는 영화',
                       '이 영화 최고'
                       ])
results
```

```python
cls=pipeline('zero-shot-classification')
cls('This course about AI Transformer',
    candidate_labels=['education','business','chemistry'])
```

```python
cls=pipeline('zero-shot-classification', model='joeddav/xlm-roberta-large-xnli')
seq='환자들은 어느 병원에 가지?'
candidate_labels=['날씨','정치','건강']
template='이 문장은 {}면이다'

cls(seq, candidate_labels, hypothesis_template=template)
```

```python
# 텍스트 생성 파이프라인
gen=pipeline('text-generation', model='distilgpt2')
gen("It's a bit warm today. What about tomorrow?", num_return_sequences=2)
```

```python
gen=pipeline('text-generation', model='skt/ko-gpt-trinity-1.2B-v0.5')
gen('skt는 구려', num_return_sequences=2)
```

```python
gen=pipeline('text-generation', model='skt/ko-gpt-trinity-1.2B-v0.5')
gen('내가 배우고 싶은 것은', max_length=100, num_return_sequences=2)
```
