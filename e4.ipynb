```python
# - Spending Score 값을 기준으로 아래와 같이 고객 등급(label)을 생성하시오.
# (if문으로 함수구현 - 함수명: def create_label(score))
# 1 ~ 33 : Low (0)
# 34 ~ 66 : Medium (1)
# 67 ~ 100 : High (2)
def create_label(score):
    labels = []
    if score>=67:
        labels.append('2')
    elif score>=34:
        labels.append('1')
    else:
        labels.append('0')
    return labels
```
